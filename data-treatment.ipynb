```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
usa_market = pd.read_csv('real-data/usa_n3538_m5993_yr2010.csv', index_col=0)
print(usa_market.shape)
usa_market.head()
```

```python
# NA analysis
# Check rows with only NA
print(usa_market.shape)
usa_market = usa_market[~usa_market.isnull().all(axis=1)]
n,m = usa_market.shape
print(usa_market.shape)
# Drop columns with more than 10% NA
usa_market = usa_market.dropna(thresh=n*0.99, axis=1)
n,m = usa_market.shape
print(usa_market.shape)
# Drop rows with more than 10% NA
usa_market = usa_market.dropna(thresh=m*0.99, axis=0)
n,m = usa_market.shape
print(usa_market.shape)
usa_market.loc[:, usa_market.isnull().any()].head()
# usa_market.isnull().sum()
```

```python
# Fill NA with KNN
from fancyimpute import KNN
usa_market_filled = pd.DataFrame(KNN(k=3).fit_transform(usa_market), columns=usa_market.columns)
usa_market_filled.head()
usa_market_filled.isnull().sum()
```

```python
# Save filled data in real-data-treated
n,m = usa_market_filled.shape
yr = 2010
usa_market_filled.index = usa_market.index
usa_market_filled.to_csv(f'real-data-treated/usa_n{n}_m{m}_yr{yr}_filled.csv', index=True)
```
